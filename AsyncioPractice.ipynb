```python
import aiohttp
import asyncio
import time

async def main():
    urls = [f"https://jsonplaceholder.typicode.com/comments/{i}" for i in range(1, 241)]
    semaphore = asyncio.Semaphore(10)
    async with aiohttp.ClientSession() as session:
        tasks = [fetch_url(session, url, semaphore) for url in urls]
        await asyncio.gather(*tasks)

async def fetch_url(session, url, semaphore):
    async with semaphore:
        response = await session.get(url)

start = time.perf_counter()
await main()
print(f"Execution time: {time.perf_counter() - start}")
    
```

```python
import random
import asyncio

queue = asyncio.Queue(7)

async def producer():
    while True:
        sleep_time = random.uniform(0.05, 0.15)
        await asyncio.sleep(sleep_time)
        event = random.randint(1,10)
        await queue.put(event)
        print(f"Event {event} added to queue. Queue size: {queue.qsize()}")

async def consumer():
    while True:
        try:
            event = await queue.get()
            print(f"Event {event} popped from queue. Queue size: {queue.qsize()}")
            async with asyncio.timeout(1):
                sleep_time = random.uniform(0.8, 1.2)
                await asyncio.sleep(sleep_time)
        except:
            print(f"Consumer timeout exceeded.")

await asyncio.gather(producer(), consumer())
```
