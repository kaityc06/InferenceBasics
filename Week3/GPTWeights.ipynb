# Prefill (512 prompt tokens)
## FLOPs calculations
Token embedding: lookup so 0 FLOPs
Position embedding: lookup so 0 FLOPs
Token + position embedding: 768 * 512 
{Layer norm: 7 * 512 * 768
Linear projections for key, value, queries: (512, 768) x (768, 2304) + (1, 2304) => 2 * 512 * 768 * 2304 + 2304 * 512
{Dot products between keys and queries: 513 * 512/2 pairwise dot products, each is 64 multiplications and 63 additions => 513 * 512/2 * (64 + 63)
Softmaxxing and scaling (512, 512) matrix => 6 * 512 * 512
Matmul between attention matrix and values => (512, 512) x (512, 64) = 2 * 512 * 512 * 64
} * 12 for each of the 12 heads
Concatenating heads: 0 FLOPs
Linear projection: (512, 768) x (768, 768) + (1, 768) = 2 * 512 * 768 * 768 + 768 * 512
Dropout: 0 FLOPs
Layer norm: 7 * 512 * 768
FFN up-projection: (512, 768) x (768, 3072) + (1, 3072) = 2 * 512 * 768 * 3072 + 512 * 3072
GELU: 8 * 512 * 3072
FFN down-projection: (512, 3072) x (3072, 768) + (1, 768) = 2 * 512 * 3072 * 768 + 512 * 768
} * 12 for each of the 12 blocks
Layer norm: 7 * 512 * 768
LM head: (512, 768) x (768, 50257) = 2 * 512 * 768 * 50257

= 1.342 × 10**11

248.9 MB

## FLOPs per bytes
539.3 FLOPs/byte
